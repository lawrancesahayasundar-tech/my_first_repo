```python
with open("sales_app.py", "w") as f:
    f.write("""
import streamlit as st
import pandas as pd

st.title("Sales Summary Dashboard")
st.subheader("Interactive view of product sales")

data = {
    "Product": ["Laptop", "Phone", "Tablet", "Headphones", "Charger"],
    "Category": ["Electronics", "Electronics", "Electronics", "Accessories", "Accessories"],
    "Sales": [50000, 30000, 20000, 10000, 5000]
}

df = pd.DataFrame(data)

category = st.sidebar.selectbox("Select Category", df["Category"].unique())
filtered_df = df[df["Category"] == category]

st.dataframe(filtered_df)
st.line_chart(filtered_df["Sales"])
""")
```
