```python
import requests

url = "https://restcountries.com/v3.1/all"

try:
    response = requests.get(url)

    # Print status code
    print("Status Code:", response.status_code)

    if response.status_code == 200:
        data = response.json()
        print("Total number of countries retrieved:", len(data))
    else:
        print("Error: Failed to fetch data. Status Code:", response.status_code)

except Exception as e:
    print("Exception occurred:", str(e))
```

```python
import requests
import pandas as pd

url = "https://restcountries.com/v3.1/all"

try:
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        data = response.json()
    else:
        print("Error fetching data")
        data = []   # fallback to avoid crash

except Exception as e:
    print("Exception:", e)
    data = []

# Now safely process data
country_list = []

for country in data:
    name = country.get("name", {}).get("common", "N/A")
    population = country.get("population", 0)
    region = country.get("region", "N/A")
    area = country.get("area", 0)

    country_list.append({
        "Country": name,
        "Population": population,
        "Region": region,
        "Area": area
    })

# Create DataFrame
df = pd.DataFrame(country_list)

# Display first 10 rows
df.head(10)
```

```python
# Clean column names (removes case/space issues)
df.columns = df.columns.str.strip().str.title()

# 1. Region with highest total population
region_population = df.groupby("Region")["Population"].sum()
highest_population_region = region_population.idxmax()
highest_population_value = region_population.max()

# 2. Country with largest area
largest_country = df.loc[df["Area"].idxmax()]

# Print results
print("Region with Highest Total Population:")
print(highest_population_region, "-", highest_population_value)

print("\nCountry with Largest Area:")
print(largest_country["Country"], "-", largest_country["Area"])
```
